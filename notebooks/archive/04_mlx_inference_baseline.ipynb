```python
LOAD = False  # set True only after you approve the model download
from narrow_ai.settings import load_settings
from narrow_ai.models.model_registry import create_backend

settings = load_settings()
settings.model.backend = "mlx"
if not LOAD:
    print("Set LOAD = True to download/load", settings.model.model_id)
else:
    backend = create_backend(settings, load_mlx=True)
    out = backend.generate([
        {"role": "system", "content": "Reply with JSON intent only."},
        {"role": "user", "content": "What can you do?"},
    ])
    print(out.text)
    print(f"load={out.load_time_ms:.0f}ms gen={out.generate_time_ms:.0f}ms")
```

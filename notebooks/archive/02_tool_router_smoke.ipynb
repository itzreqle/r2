# 02 — Tool router smoke

Fake backend → orchestrator → allowlisted tools.


```python
from pathlib import Path
from narrow_ai.assistant.orchestrator import AssistantOrchestrator
from narrow_ai.models.fake_backend import FakeModelBackend
from narrow_ai.memory.preferences import PreferenceStore
from narrow_ai.settings import Settings

ws = Path(".").resolve()
settings = Settings()
settings.workspace.path = str(ws)
prefs = PreferenceStore(Path("/tmp/narrow-ai-nb-prefs.db"))
orch = AssistantOrchestrator(settings, FakeModelBackend(), preferences=prefs)
for q in ["What can you do?", "Check the current Git status.", "Find Python files in the project."]:
    r = orch.handle_text(q)
    print("Q:", q)
    print(" intent:", r.decision.intent if r.decision else None)
    print(" ", r.text[:200].replace("\n", " | "))
    print()
```

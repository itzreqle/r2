# 01 — Intent schemas

Round-trip validation and rejection of unsafe patterns.


```python
from narrow_ai.intents.classifier import parse_intent_decision
from narrow_ai.intents.schemas import SearchFilesParams
from pydantic import ValidationError

d = parse_intent_decision('{"intent":"git_status","parameters":{}}')
print(d)
try:
    SearchFilesParams(pattern="*.py; rm -rf /")
except ValidationError as e:
    print("blocked:", e.error_count(), "errors")
```

```python
from narrow_ai.intents.registry import default_registry
from narrow_ai.tools.permissions import PermissionGate
from narrow_ai.settings import Settings
from narrow_ai.intents.schemas import IntentDecision, IntentName

gate = PermissionGate(Settings(), default_registry())
print(gate.check(IntentDecision(intent=IntentName.GIT_STATUS, parameters={})))
print(gate.check(IntentDecision(intent=IntentName.RUN_TESTS, parameters={"command_id":"x"})))
```

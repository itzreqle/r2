# 05 — Dataset EDA


```python
from pathlib import Path
import json
from collections import Counter

root = Path("..") if Path("data").exists() is False else Path(".")
# notebooks often run with cwd=notebooks/
data_dir = Path("../data") if Path("../data").exists() else Path("data")
counts = Counter()
for split in ("train", "valid", "test"):
    path = data_dir / f"{split}.jsonl"
    if not path.exists():
        print("missing", path, "— run: uv run python training/prepare_data.py")
        continue
    n = 0
    for line in path.read_text().splitlines():
        if not line.strip():
            continue
        n += 1
        obj = json.loads(line)
        asst = next(m["content"] for m in obj["messages"] if m["role"]=="assistant")
        intent = json.loads(asst).get("intent", "?")
        counts[f"{split}:{intent}"] += 1
    print(split, n)
print(dict(counts))
```
